```python
f_tombamento_ativo = {
    "TOMB-901": "Ativo_01",
    "TOMB-902": "Ativo_02",
    "TOMB-903": "Ativo_03"
}

f_inversa = {ativo: tombamento for tombamento, ativo in f_tombamento_ativo.items()}
print(f"Função Inversa: {f_inversa}")

valores_distintos = len(set(f_tombamento_ativo.values())) == len(f_tombamento_ativo.keys())

bijetora = valores_distintos
print(f"A função é bijetora? {'Sim' if bijetora else 'Não'}")
```

```python

mapa_f = {"TOMB-901": "Ativo_01", "TOMB-902": "Ativo_02"}
mapa_g = {"Ativo_01": "Educação", "Ativo_02": "Saúde"}

def f(tombamento):
    ativo = mapa_f.get(tombamento)
    if not ativo:
        raise ValueError(f"Tombamento '{tombamento}' não existe.")
    return ativo


def g(ativo):
    secretaria = mapa_g.get(ativo)
    if not secretaria:
        raise ValueError(f"Ativo '{ativo}' sem secretaria vinculada.")
    return secretaria

def pipeline_composicao(x):
    try:
        resultado = g(f(x))
        return f"Sucesso! O tombamento {x} pertence à Secretaria de {resultado}."
    except ValueError as erro:
        return f"Erro de Robustez: {erro}"

print(pipeline_composicao("TOMB-901"))
print(pipeline_composicao("TOMB-999"))
```
